# Grundelemente - Variablen

Variablen sind so etwas wie 'Schubladen' oder 'Behälter' für Werte.
Eine Variable hat einen Variablennamen und einen Inhalt. Ich kann etwa eine Variable mit dem Namen '*meine_Variable*' erzeugen und dieser Variablen einen Wert '*52*' zuweisen.

In Python sieht das dann so aus:

```python
meine_variable = 52
```

Der Python Interpreter merkt sich nun, dass in der 'Schublade' *meine_variable* der Wert 52 gespeichert ist.
Ich kann den Inhalt der Variable jederzeit abfragen und ausgeben:

```python
print (meine_variable)
```

oder ich kann der Variablen neue Werte zuweisen, die Variable ändern, neue Variablen erzeugen oder die Variable löschen:

```python
meine_variable = 12

a = 2
b = a * meine_variable

del meine_variable

print (a)
print (b)
print (meine_variable)
```

Die Werte in Variablen können auch Texte:

```python
text_variable = "Hallo Welt!"
print(text_variable)
```

oder sogar ganze Listen von Werten sein. Bei solchen Listen kann ich auch auf die einzelnen Listenelemente zugreifen.

```python
list_variable = [1,2,4,8,16,32,64,128,256]
print (list_variable)
print (list_variable[0])
print (list_variable[2:4])
```

# Grundelemente - Bedingungen
Mit Hilfe von Bedingungen kann ich Werte von Variablen überprüfen und darauf reagieren.
In Python wird dafür das **if** verwendet.
Möchte ich prüfen, ob eine Zahlenvariable größer oder kleiner als 0 ist:


```python
zahl = 100

if zahl > 0:
  print ("Zahl ist positiv")
```

Mit den Schlüsselwörtern **elif** und **else** lassen sich mehr Varianten prüfen

```python
zahl = 0

if zahl > 0:
  print ("Zahl ist positiv")
elif zahl < 0:
  print ("Zahl ist negativ")
else:
  print ("Zahl ist null")
```

# Grundelemente - Schleifen #
Mit Schleifen lassen sich Codeelemente wiederholen. Bei Python gibt es 2 Schleifentypen:
* die **while** Schleife wird solange ausgeführt, wie eine Bedingung wahr ist:

```python
zahl = 1
while zahl < 300:
  print (zahl)
  zahl = zahl * 2
```

* die **for** Schleife nimmt alle Elemente eines *sequentiellen* Wertes und bearbeitet diese einzeln

```python
farben = ["rot", "orange", "gelb", "grün", "blau", "violett"]
for einzelfarbe in farben:
  print(einzelfarbe)
```

# Codebeispiele #
Berechnung der Fibonacci Zahlen - siehe [Wikipedia](https://de.wikipedia.org/wiki/Fibonacci-Folge)

```python
# Fibonacci Zahlen
# Berechne die ersten 20 Fibonacci Zahlen
i = 1
zahl_1 = 1
zahl_2 = 1
while i < 20:
  fibo = zahl_1 + zahl_2
  print(fibo)
  zahl_1 = zahl_2
  zahl_2 = fibo
  i = i + 1
```

Buchstaben ersetzen in einem Text

```python
text = "Mähmaschinen mähen besser als Nähmaschinen"

for buchstabe in text:
  if buchstabe == "ä":
    buchstabe = "ae"
  print (buchstabe, end='')
```

Ermittle den größten gemeinsamen Teiler - Algorithmus von Euklid

```python
zahl_1 = 89382312435
zahl_2 = 23423535
rest = -1
while rest != 0:            # solange die Schleife laufen lassen, wie rest ungleich 0
  rest = zahl_1 % zahl_2    # das ist der sog. modulo, welcher den Rest einer Divion liefert
  zahl_1 = zahl_2
  zahl_2 = rest
print (zahl_1)
```
